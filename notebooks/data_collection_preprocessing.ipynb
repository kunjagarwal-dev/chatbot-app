```python
with open("../data/raw/movie_lines.txt", encoding="iso-8859-1") as f:
    lines = f.readlines()

print(f"Total lines: {len(lines)}")
for line in lines[:5]:
    print(line)
```

```python
with open("../data/raw/movie_conversations.txt", encoding="iso-8859-1") as f:
    conversations = f.readlines()

print(f"Total conversations: {len(conversations)}")
for conv in conversations[:5]:
    print(conv)
```

```python
lines[0].split(" +++$+++ ")
```

```python
line_dict = {}
for line in lines:
    parts = line.split(" +++$+++ ")
    if len(parts) == 5:
        line_id = parts[0]
        text = parts[4].strip()
        line_dict[line_id] = text

print(f"Total unique lines: {len(line_dict)}")
```

```python
import ast

conversation_list = []
for conv in conversations:
    parts = conv.split(" +++$+++ ")
    if len(parts) == 4:
        line_ids_str = parts[3].strip()
        line_ids = ast.literal_eval(line_ids_str)
        conversation_list.append(line_ids)

print(f"Total conversations: {len(conversation_list)}")
```

```python
pairs = []
for line_ids in conversation_list:
    for i in range(len(line_ids)-1):
        input_line = line_dict.get(line_ids[i], "")
        target_line = line_dict.get(line_ids[i+1], "")
        if input_line and target_line:
            pairs.append((input_line, target_line))
        
print(f"Total pairs: {len(pairs)}")
pairs[:5]
```

```python
for i in range(len(pairs)):
    input_word_count = len(pairs[i][0].split())
    target_word_count = len(pairs[i][1].split())
    print(f"Pair {i+1}: Input word count: {input_word_count}, Target word count: {target_word_count}")
```

```python
print(f"Max input word count: {max(len(pair[0].split()) for pair in pairs)}")
print(f"Max output word count: {max(len(pair[1].split()) for pair in pairs)}")
```

```python
import numpy as np

input_lengths = [len(pair[0].split()) for pair in pairs]
target_lengths = [len(pair[1].split()) for pair in pairs]

for name, lengths in [("Input", input_lengths), ("Target", target_lengths)]:
    print(f"{name} — mean: {np.mean(lengths):.1f}, median: {np.median(lengths):.1f}, "
          f"95th percentile: {np.percentile(lengths, 95):.1f}, max: {max(lengths)}")
```

```python
filtered_pairs = [pair for pair in pairs if len(pair[0].split()) < 32 and len(pair[1].split()) < 32]        
```

```python
len(filtered_pairs)
```

```python
tokensized_pairs = [(pair[0].split(), pair[1].split()) for pair in filtered_pairs]
tokensized_pairs[:5]
```

```python
from collections import Counter

token_counts = Counter(
    token for pair in tokensized_pairs for token in pair[0] + pair[1]
)

len(token_counts)
```

```python
singletons = sum(1 for word, count in token_counts.items() if count == 1)
rare_words = sum(1 for word, count in token_counts.items() if count <= 5)

print(f"Words appearing exactly once: {singletons}")
print(f"Words appearing 5 or fewer times: {rare_words}")
print(f"Words appearing more than 5 times (kept vocab): {len(token_counts) - rare_words}")
```

```python
# Build vocabulary from token_counts, keeping only words with count > 5

MIN_COUNT = 5

# Filter to frequent words only
kept_words = [word for word, count in token_counts.items() if count > MIN_COUNT]

# Special tokens first, in a fixed order
special_tokens = ["<PAD>", "<START>", "<END>", "<UNK>"]

# Build the full vocabulary list: special tokens + frequent words
vocab_list = special_tokens + sorted(kept_words)

# Build word -> index and index -> word mappings
word2idx = {word: idx for idx, word in enumerate(vocab_list)}
idx2word = {idx: word for word, idx in word2idx.items()}

print(f"Vocabulary size (including special tokens): {len(word2idx)}")

# Quick sanity checks
print(f"Index of <PAD>: {word2idx['<PAD>']}")
print(f"Index of <UNK>: {word2idx['<UNK>']}")
print(f"Index of 'hello' (if present): {word2idx.get('hello', 'NOT IN VOCAB')}")

# Helper function: convert a sentence into token IDs, using <UNK> for unknown words
def sentence_to_ids(sentence, word2idx):
    return [word2idx.get(word, word2idx["<UNK>"]) for word in sentence.split()]

# Test it on a sample pair
sample_input, sample_target = filtered_pairs[0]
print(f"Sample input: {sample_input}")
print(f"Encoded: {sentence_to_ids(sample_input, word2idx)}")
```

```python
filtered_pairs[:5]
```

```python
encoder_input = []
decoder_input = []
decoder_output = []

for pair in filtered_pairs:
    input_text = pair[0]
    target_text = pair[1]

    encoder_input.append(sentence_to_ids(input_text, word2idx))
    decoder_input.append([word2idx["<START>"]] + sentence_to_ids(target_text, word2idx))
    decoder_output.append(sentence_to_ids(target_text, word2idx) + [word2idx["<END>"]])

print(f"Sample pair: {filtered_pairs[0]}")
print(f"Encoder input IDs: {encoder_input[0]}")
print(f"Decoder input IDs: {decoder_input[0]}")
print(f"Decoder output IDs: {decoder_output[0]}")
```

```python
print(f"Total encoder input sequences: {len(encoder_input)}")
```

```python
import tensorflow
from tensorflow import keras
from keras.preprocessing.sequence import pad_sequences

MAX_SEQ_LEN = 33

encoder_input = pad_sequences(encoder_input, maxlen=MAX_SEQ_LEN, padding="post")
decoder_input = pad_sequences(decoder_input, maxlen=MAX_SEQ_LEN, padding="post")
decoder_output = pad_sequences(decoder_output, maxlen=MAX_SEQ_LEN, padding="post")

(len(encoder_input), len(decoder_input), len(decoder_output))
```

```python
import joblib
import numpy as np

# Save the padded numpy arrays
np.save("../data/processed/encoder_input.npy", encoder_input)
np.save("../data/processed/decoder_input.npy", decoder_input)
np.save("../data/processed/decoder_output.npy", decoder_output)

# Save the vocabulary mappings
joblib.dump(word2idx, "../data/processed/word2idx.pkl")
joblib.dump(idx2word, "../data/processed/idx2word.pkl")

print("All Day 1 artifacts saved to data/processed/")
```
